```python
import os
os.chdir("/home/anis/cluster_security")  # adjust to your actual path
print(os.getcwd())
print(os.path.exists("ground_truth_log.csv"))

import sys, json, glob
import pandas as pd
from datetime import datetime, timezone
sys.path.append("remediation-engine")

sys.path.append("../remediation-engine")  # wherever features.py lives
from features import extract_features, prune_window, FEATURE_COLUMNS, WINDOW_SECONDS

GT_LOG = "ground_truth_log.csv"
ALERT_FILES = ["remediation-engine/falco_alerts.jsonl"]
 
print(ALERT_FILES)
```

```python
gt_cols = ["run_id","start_utc","end_utc","label","type","pod","expected_rule","detail"]
gt = pd.read_csv(GT_LOG, header=None, names=gt_cols, skiprows=1)

# legacy v5 rows have no run_id (3-col schema) -> NaN after reindex; drop them
gt = gt.dropna(subset=["run_id"]).copy()

gt["start_dt"] = pd.to_datetime(gt["start_utc"], format="%Y-%m-%dT%H:%M:%S.%f", utc=True, errors="coerce")
gt["end_dt"]   = pd.to_datetime(gt["end_utc"], format="%Y-%m-%dT%H:%M:%S.%f", utc=True, errors="coerce")
gt = gt.dropna(subset=["start_dt", "end_dt"])

print(gt.shape)
gt.head()
```

```python
pod_created = gt[gt["type"] == "pod_created"][["run_id","pod","start_dt"]]
pod_created = pod_created.rename(columns={"start_dt": "created_at"})
pod_created_map = dict(zip(pod_created["pod"], pod_created["created_at"]))
pod_created_map
```

```python
# Adjust field names here if your webhook JSON differs.
rows = []
for path in ALERT_FILES:
    with open(path) as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            try:
                a = json.loads(line)
            except json.JSONDecodeError:
                continue
            fields = a.get("output_fields", {})
            pod = fields.get("k8s.pod.name")
            ns = fields.get("k8s.ns.name") or "default"
            rule = a.get("rule")
            priority = a.get("priority")
            ts_raw = a.get("time") or a.get("output_fields", {}).get("evt.time.iso8601")
            if not pod or not rule or not ts_raw:
                continue
            rows.append({"pod": pod, "namespace": ns, "rule": rule,
                         "priority": priority, "ts_raw": ts_raw})

alerts = pd.DataFrame(rows)
alerts["ts"] = pd.to_datetime(alerts["ts_raw"], utc=True, errors="coerce")
alerts = alerts.dropna(subset=["ts"])
print(alerts.shape)
alerts.head()
```

```python
def label_alert(pod, ts):
    """Look up pod's rows in gt, check if ts falls in [start-1s, end+3s]
    of any attack/benign action. Return (label, run_id) or (None, None)
    if unlabeled (excluded from metrics)."""
    candidates = gt[(gt["pod"] == pod) & (gt["label"].isin(["attack", "benign"]))]
    for _, row in candidates.iterrows():
        window_start = row["start_dt"] - pd.Timedelta(seconds=1)
        window_end = row["end_dt"] + pd.Timedelta(seconds=3)
        if window_start <= ts <= window_end:
            return row["label"], row["run_id"]
    return None, None

alerts[["label", "run_id"]] = alerts.apply(
    lambda r: pd.Series(label_alert(r["pod"], r["ts"])), axis=1
)
alerts["label"].value_counts(dropna=False)
```

```python
training_rows = []

for pod, g in alerts.groupby("pod"):
    g = g.sort_values("ts")
    events = list(zip(g["ts"], g["rule"], g["priority"]))  # full pod history for windowing

    for i, row in g.iterrows():
        if pd.isna(row["label"]):
            continue  # unlabeled, skip

        t = row["ts"]
        # trailing window: all events on this pod within WINDOW_SECONDS before t
        window_events = [
            (e_ts.timestamp(), e_rule, e_prio)
            for (e_ts, e_rule, e_prio) in events
            if 0 <= (t - e_ts).total_seconds() <= WINDOW_SECONDS
        ]

        created_at = pod_created_map.get(pod)
        pod_age = (t - created_at).total_seconds() if created_at is not None else None

        feat = extract_features(window_events, pod_age_seconds=pod_age)
        feat["pod"] = pod
        feat["run_id"] = row["run_id"]
        feat["timestamp"] = t
        feat["label"] = 1 if row["label"] == "attack" else 0
        training_rows.append(feat)

train_df = pd.DataFrame(training_rows)
print(train_df.shape)
train_df.head()
```

```python
print("Class balance:\n", train_df["label"].value_counts())
print("\nPer run_id:\n", train_df.groupby("run_id")["label"].value_counts())
print("\nRows with all-zero features (check these aren't dominating):",
      (train_df[FEATURE_COLUMNS].sum(axis=1) == 0).sum())

train_df.to_csv("training_table.csv", index=False)
```
