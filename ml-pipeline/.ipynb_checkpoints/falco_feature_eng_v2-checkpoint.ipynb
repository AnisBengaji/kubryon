```python
import json
import glob
import pandas as pd

# Explicitly exclude known-polluted files, don't just glob everything
exclude = ["falco_alerts_polluted_backup.jsonl", "falco_alerts_polluted_backup2.jsonl"]
all_files = [f for f in glob.glob("*.jsonl") if f not in exclude]
print("Files found:", all_files)

all_raw_events = []
for fname in all_files:
    with open(fname) as f:
        events = [json.loads(line) for line in f if line.strip()]
        all_raw_events.extend(events)
        print(f"  {fname}: {len(events)} events")

print(f"\nTotal combined (before dedup): {len(all_raw_events)} events")
```

```python
seen_uuids = set()
deduped_events = []
for e in all_raw_events:
    uuid = e.get("uuid")
    if uuid and uuid not in seen_uuids:
        seen_uuids.add(uuid)
        deduped_events.append(e)

print(f"Total after deduplication: {len(deduped_events)} events")
raw_events = deduped_events  # use this going forward, replacing your original raw_events variable
```

```python
def flatten_event(e):
    fields = e.get("output_fields", {})
    return {
        "time": e.get("time"),
        "rule": e.get("rule"),
        "priority": e.get("priority"),
        "pod_name": fields.get("k8s.pod.name"),
        "namespace": fields.get("k8s.ns.name"),
        "proc_name": fields.get("proc.name"),
        "container_id": fields.get("container.id"),
    }

df_raw = pd.DataFrame([flatten_event(e) for e in raw_events])
df_raw["time"] = pd.to_datetime(df_raw["time"])
df_raw.head(20)
```

```python
print("=== Rule counts ===")
print(df_raw["rule"].value_counts())
print()

print("=== Priority counts ===")
print(df_raw["priority"].value_counts())
print()

print("=== Pod name counts (NaN = missing metadata) ===")
print(df_raw["pod_name"].value_counts(dropna=False))
print()

print("=== Process name counts ===")
print(df_raw["proc_name"].value_counts(dropna=False))
print()

print(f"Rows with missing pod_name: {df_raw['pod_name'].isna().sum()} / {len(df_raw)}")
print(f"Rows with container_id == 'host': {(df_raw['container_id'] == 'host').sum()}")
```

```python
# Investigate the new/unexpected rule
mystery = df_raw[df_raw["rule"] == "Directory traversal monitored file read"]
print(mystery)
print()

# Investigate remaining host-level events
host_events = df_raw[df_raw["container_id"] == "host"]
print(host_events)
```

```python
before = len(df_raw)
df_clean = df_raw[df_raw["pod_name"].notna()]
print(f"Dropped {before - len(df_clean)} rows with missing pod_name (duplicate <NA> copies)")

before = len(df_clean)
df_clean = df_clean[df_clean["container_id"] != "host"]
print(f"Dropped {before - len(df_clean)} rows from host-level noise")

before = len(df_clean)
df_clean = df_clean[df_clean["rule"] != "Falco internal: syscall event drop"]
print(f"Dropped {before - len(df_clean)} internal Falco system events")

# Exclude Falco/Sidekick's own infrastructure pods. These aren't attack
# targets -- they just run continuously in the background for the whole
# collection period, and were dominating the "normal" class with thousands
# of irrelevant empty windows once resampled.
infra_pod_prefixes = ("falco-", "redisearch")
before = len(df_clean)
df_clean = df_clean[~df_clean["pod_name"].str.startswith(infra_pod_prefixes)]
print(f"Dropped {before - len(df_clean)} rows from non-target infrastructure pods")

print(f"\nFinal clean dataset: {len(df_clean)} rows")
print(f"Remaining pods: {sorted(df_clean['pod_name'].unique())}")
df_clean["rule"].value_counts()
```

```python
def build_features(df, window_seconds=60, session_gap_minutes=10):
    df = df.set_index("time").sort_index()

    priority_weight = {
        "Critical": 4, "Error": 3, "Warning": 2, "Notice": 1, "Informational": 0, "Debug": 0
    }
    df["priority_weight"] = df["priority"].map(priority_weight).fillna(0)

    # Build known_rules directly from what's actually in the data instead of
    # hand-typing names that can silently drift out of sync with Falco's
    # real rule names. This is what caused "suspicious_suid_bit_set" to be
    # all zeros before -- the real rule is "Set Setuid or Setgid bit".
    known_rules = sorted(df["rule"].unique())

    features = []
    for pod_name, pod_df in df.groupby("pod_name"):
        pod_df = pod_df.sort_index()

        # Split this pod's events into separate *sessions* wherever there's
        # a large time gap (e.g. the multi-day gap between collection runs).
        # Without this, resample() spans a pod's full first-to-last
        # timestamp range -- for a persistent pod name reused across
        # sessions (like test-pod), that generates ~2 days of empty 60s
        # windows just because two runs happened to reuse the same name.
        gaps = pod_df.index.to_series().diff() > pd.Timedelta(minutes=session_gap_minutes)
        session_id = gaps.cumsum()

        for _, session_df in pod_df.groupby(session_id):
            resampled = session_df.resample(f"{window_seconds}s")
            for window_start, window_df in resampled:
                row = {
                    "window_start": window_start,
                    "pod_name": pod_name,
                    "event_count": len(window_df),
                    "critical_count": (window_df["priority"] == "Critical").sum(),
                    "distinct_rules": window_df["rule"].nunique(),
                    "max_priority_weight": window_df["priority_weight"].max() if len(window_df) else 0,
                    "avg_priority_weight": window_df["priority_weight"].mean() if len(window_df) else 0,
                }
                rules_present = set(window_df["rule"])
                for rule_name in known_rules:
                    col_name = rule_name.lower().replace(" ", "_").replace("/", "_")[:30]
                    row[col_name] = int(rule_name in rules_present)
                features.append(row)

    return pd.DataFrame(features)

feature_df = build_features(df_clean, window_seconds=60)
print(f"Total feature rows: {len(feature_df)}")
print(f"Normal (empty) windows: {(feature_df['event_count'] == 0).sum()}")
print(f"Active (non-empty) windows: {(feature_df['event_count'] > 0).sum()}")
feature_df.head(20)
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

base_columns = [
    "event_count", "critical_count", "distinct_rules",
    "max_priority_weight", "avg_priority_weight",
]
rule_columns = [c for c in feature_df.columns if c not in base_columns + ["window_start", "pod_name"]]
feature_columns = base_columns + rule_columns

X = feature_df[feature_columns].fillna(0)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# NOTE: re-check this contamination value now that infra pods are excluded
# and sessions are split. 0.13 was tuned against a baseline that was ~99%
# empty windows from continuously-running infra pods -- that ratio no
# longer reflects the real normal/active split in this dataset.
model = IsolationForest(
    n_estimators=100,
    contamination=0.13,
    random_state=42
)
model.fit(X_scaled)

feature_df["anomaly_score"] = model.decision_function(X_scaled)
feature_df["is_anomaly"] = model.predict(X_scaled)  # -1 = anomaly, 1 = normal

print(feature_df[feature_df["is_anomaly"] == -1][["window_start", "pod_name", "event_count", "critical_count", "anomaly_score"]])
```

```python
anomalies = feature_df[feature_df["is_anomaly"] == -1].sort_values("anomaly_score")
print(anomalies[["window_start", "pod_name", "event_count", "critical_count", "distinct_rules", "anomaly_score"]])
```
