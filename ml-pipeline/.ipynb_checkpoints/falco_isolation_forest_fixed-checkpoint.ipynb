# Fixed Falco Kubernetes anomaly-detection model

This notebook rebuilds the project's Isolation Forest pipeline while fixing the problems found during evaluation.

**Key fixes**
- 120-second windows, matching the remediation engine.
- `test-pod` and `test-pod-apt` are completely held out from training.
- Robust Falco parsing and duplicate handling.
- Infrastructure/security namespaces excluded.
- Priority-aware features.
- Counts for all relevant Falco security rules instead of five binary indicators.
- Severity, rate, diversity, process, and container features.
- Ground truth matched by **pod + timestamp window**.
- `attack` / `benign` / `unknown` labels; unknown is never silently treated as benign.
- Production thresholds are recalibrated for the new model instead of reusing the old `-0.05/-0.20` thresholds.
- A portable `falco_isolation_forest_fixed.joblib` bundle is exported.

The notebook expects the Falco `*.jsonl` files and `ground_truth_log.csv` in the same directory as the notebook.


```python
# 1. Configuration
from pathlib import Path
import json, re, warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

WINDOW_SECONDS = 120
TEST_PODS = {"test-pod", "test-pod-apt"}
DATA_DIR = Path(".")
GT_PATH = DATA_DIR / "ground_truth_log.csv"
RAW_FILES = sorted(p for p in DATA_DIR.glob("*.jsonl")
                   if "remediation" not in p.name.lower())

EXCLUDED_NAMESPACES = {
    "kube-system", "kube-public", "kube-node-lease",
    "falco", "falco-ui", "monitoring", "local-path-storage",
}
PRIORITY_WEIGHT = {
    "Debug": 0, "Informational": 0, "Notice": 3,
    "Warning": 2, "Error": 3, "Critical": 4,
}
SCORABLE_PRIORITIES = {"Notice", "Warning", "Error", "Critical"}

print("JSONL files:", [p.name for p in RAW_FILES])
print("Ground truth:", GT_PATH)
print("Test pods:", sorted(TEST_PODS))
```

```python
# 2. Load and deduplicate raw Falco events
if not RAW_FILES:
    raise FileNotFoundError("No *.jsonl files found beside this notebook.")
if not GT_PATH.exists():
    raise FileNotFoundError("ground_truth_log.csv not found beside this notebook.")

raw_events = []
for path in RAW_FILES:
    n = 0
    with path.open("r", encoding="utf-8") as f:
        for line in f:
            if not line.strip():
                continue
            try:
                raw_events.append(json.loads(line))
                n += 1
            except json.JSONDecodeError:
                pass
    print(f"{path.name}: {n} valid JSON records")

print("Before UUID dedup:", len(raw_events))

seen = set()
deduped = []
for e in raw_events:
    uid = e.get("uuid")
    if uid:
        if uid in seen:
            continue
        seen.add(uid)
    deduped.append(e)

print("After UUID dedup:", len(deduped))
```

```python
# 3. Flatten, clean, and filter Falco events
def first_nonempty(*values):
    for v in values:
        if v is not None and str(v).strip().lower() not in {"", "nan", "<na>", "none"}:
            return v
    return None

def flatten_event(e):
    fields = e.get("output_fields") or {}
    return {
        "time": e.get("time"),
        "rule": e.get("rule"),
        "priority": e.get("priority"),
        "pod_name": first_nonempty(
            fields.get("k8s.pod.name"),
            fields.get("k8smeta.pod.name"),
            fields.get("container.name"),
        ),
        "namespace": first_nonempty(
            fields.get("k8s.ns.name"),
            fields.get("k8smeta.ns.name"),
        ),
        "proc_name": first_nonempty(fields.get("proc.name")),
        "container_id": first_nonempty(fields.get("container.id")),
    }

df = pd.DataFrame([flatten_event(e) for e in deduped])
df["time"] = pd.to_datetime(df["time"], utc=True, errors="coerce")
df = df[df["time"].notna() & df["pod_name"].notna()].copy()

df = df[df["container_id"].fillna("").astype(str).str.lower().ne("host")]
df = df[~df["namespace"].fillna("").isin(EXCLUDED_NAMESPACES)]
df["priority"] = df["priority"].fillna("Unknown")
df["rule"] = df["rule"].fillna("Unknown")
df = df[df["priority"].isin(SCORABLE_PRIORITIES)].copy()

df["priority_weight"] = df["priority"].map(PRIORITY_WEIGHT).fillna(0.0)

# Protect against duplicate captures with different UUIDs.
fp = ["time", "rule", "priority", "pod_name", "proc_name", "container_id"]
df = df.drop_duplicates(subset=fp).sort_values("time").reset_index(drop=True)

print("Clean scorable events:", len(df))
print("Pods:", df["pod_name"].nunique())
print("Rules:", df["rule"].nunique())
print("Range:", df["time"].min(), "->", df["time"].max())
display(df.head())
```

```python
# 4. Load and parse ground truth
gt = pd.read_csv(GT_PATH)
required = {"timestamp_utc", "label", "detail"}
missing = required - set(gt.columns)
if missing:
    raise ValueError(f"Missing ground-truth columns: {missing}")

gt["timestamp_utc"] = pd.to_datetime(gt["timestamp_utc"], utc=True, errors="coerce")
gt["label"] = gt["label"].astype(str).str.strip()
gt["detail"] = gt["detail"].astype(str).str.strip()

gt["gt_pod"] = gt["detail"].str.split(":", n=1).str[-1].str.strip()
attacks = gt[(gt["label"] == "attack") & gt["timestamp_utc"].notna()].copy()

idle_intervals = []
open_start = None
for _, r in gt.sort_values("timestamp_utc").iterrows():
    if r["label"] == "idle_start":
        open_start = r["timestamp_utc"]
    elif r["label"] == "idle_end" and open_start is not None:
        idle_intervals.append((open_start, r["timestamp_utc"]))
        open_start = None

print(gt["label"].value_counts())
print("\nIdle intervals:")
for a, b in idle_intervals:
    print(a, "->", b)
```

## Feature engineering

The previous model used only five binary rule-presence indicators. The new representation uses rule **counts** plus aggregate behavior/severity features. This makes behaviors such as package management, environment-variable privilege escalation, shell-history deletion, SSH-information access, and root writes visible to the model instead of collapsing them into generic event counts.


```python
# 5. Fixed security-rule vocabulary from the project data
SECURITY_RULES = [
    "Terminal shell in container",
    "Drop and execute new binary in container",
    "Read sensitive file untrusted",
    "Suspicious SUID bit set",
    "Contact K8S API Server From Container",
    "Set Setuid or Setgid bit",
    "Write below root",
    "Launch Package Management Process in Container",
    "Read ssh information",
    "Potential Local Privilege Escalation via Environment Variables Misuse",
    "Delete or rename shell history",
    "Launch Ingress Remote File Copy Tools in Container",
    "Unexpected UDP Traffic",
    "Change thread namespace",
    "Change namespace privileges via unshare",
    "Launch Privileged Container",
    "Schedule Cron Jobs",
    "Launch Sensitive Mount Container",
    "Basic Interactive Reconnaissance",
]

RULE_FEATURES = {
    rule: "rule_" + re.sub(r"[^a-z0-9]+", "_", rule.lower()).strip("_")[:55]
    for rule in SECURITY_RULES
}
RULE_FEATURE_COLUMNS = list(RULE_FEATURES.values())

def build_window(events, window_start, pod_name, rule_features=RULE_FEATURES):
    end = window_start + pd.Timedelta(seconds=WINDOW_SECONDS)
    w = events[
        events["pod_name"].eq(pod_name)
        & events["time"].ge(window_start)
        & events["time"].lt(end)
    ]

    n = len(w)
    row = {
        "window_start": window_start,
        "pod_name": pod_name,
        "event_count": n,
        "notice_count": int((w["priority"] == "Notice").sum()),
        "warning_count": int((w["priority"] == "Warning").sum()),
        "error_count": int((w["priority"] == "Error").sum()),
        "critical_count": int((w["priority"] == "Critical").sum()),
        "security_event_count": n,
        "distinct_rules": int(w["rule"].nunique()) if n else 0,
        "distinct_processes": int(w["proc_name"].nunique()) if n else 0,
        "distinct_containers": int(w["container_id"].nunique()) if n else 0,
        "max_priority_weight": float(w["priority_weight"].max()) if n else 0.0,
        "avg_priority_weight": float(w["priority_weight"].mean()) if n else 0.0,
        "total_priority_weight": float(w["priority_weight"].sum()) if n else 0.0,
    }

    counts = w["rule"].value_counts()
    for rule, col in rule_features.items():
        row[col] = int(counts.get(rule, 0))

    row["events_per_minute"] = n / (WINDOW_SECONDS / 60.0)
    row["critical_rate"] = row["critical_count"] / max(n, 1)
    row["high_severity_rate"] = (
        row["critical_count"] + row["error_count"] + row["warning_count"]
    ) / max(n, 1)
    return row
```

```python
# 6. Build 120-second active windows
df["window_start"] = df["time"].dt.floor(f"{WINDOW_SECONDS}s")

rows = []
for (window_start, pod_name), _ in df.groupby(["window_start", "pod_name"], sort=True):
    rows.append(build_window(df, window_start, pod_name))

feature_df = pd.DataFrame(rows).sort_values(
    ["window_start", "pod_name"]
).reset_index(drop=True)

print("Active feature windows:", len(feature_df))
display(feature_df.head())
```

```python
# 7. Correct ground-truth alignment
def label_window(row):
    start = row["window_start"]
    end = start + pd.Timedelta(seconds=WINDOW_SECONDS)
    pod = str(row["pod_name"]).strip()

    # Attack wins if any attack annotation for this exact pod is inside the window.
    hit = attacks[
        attacks["gt_pod"].eq(pod)
        & attacks["timestamp_utc"].ge(start)
        & attacks["timestamp_utc"].lt(end)
    ]
    if not hit.empty:
        return "attack"

    # Benign is allowed only for a complete documented idle interval.
    for idle_start, idle_end in idle_intervals:
        if start >= idle_start and end <= idle_end:
            return "benign"

    return "unknown"

feature_df["ground_truth"] = feature_df.apply(label_window, axis=1)

print(feature_df["ground_truth"].value_counts(dropna=False))
```

```python
# 8. Held-out test split
train_df = feature_df[~feature_df["pod_name"].isin(TEST_PODS)].copy()
test_df = feature_df[feature_df["pod_name"].isin(TEST_PODS)].copy()

# Never train on a window explicitly labelled as attack.
train_df = train_df[train_df["ground_truth"] != "attack"].copy()

print("Training windows:", len(train_df))
print("Test windows:", len(test_df))
print("Test pods:", sorted(test_df["pod_name"].unique()))
```

```python
# 9. Model feature matrix
BASE_FEATURES = [
    "event_count", "notice_count", "warning_count", "error_count",
    "critical_count", "security_event_count", "distinct_rules",
    "distinct_processes", "distinct_containers", "max_priority_weight",
    "avg_priority_weight", "total_priority_weight", "events_per_minute",
    "critical_rate", "high_severity_rate",
]
FEATURE_COLUMNS = BASE_FEATURES + RULE_FEATURE_COLUMNS

X_train = train_df[FEATURE_COLUMNS].astype(float).fillna(0.0)
X_test = test_df[FEATURE_COLUMNS].astype(float).fillna(0.0)

print("Total features:", len(FEATURE_COLUMNS))
print("Rule-count features:", len(RULE_FEATURE_COLUMNS))
```

```python
# 10. Train Isolation Forest
from sklearn.ensemble import IsolationForest

model = IsolationForest(
    n_estimators=500,
    contamination="auto",
    max_samples="auto",
    max_features=1.0,
    random_state=42,
    n_jobs=-1,
)
model.fit(X_train)

train_df["anomaly_score"] = model.decision_function(X_train)
test_df["anomaly_score"] = model.decision_function(X_test)
train_df["native_prediction"] = model.predict(X_train)
test_df["native_prediction"] = model.predict(X_test)

print("Training score range:",
      round(train_df["anomaly_score"].min(), 6),
      "to",
      round(train_df["anomaly_score"].max(), 6))
```

## Production thresholds

The old `-0.05` and `-0.20` thresholds belonged to the previous model/feature distribution. They should not be copied to a retrained Isolation Forest.

The thresholds below are derived only from the normal training score distribution:
- isolate: bottom 2%
- kill: bottom 0.5%

They do **not** use the held-out attack labels.


```python
# 11. Calibrate action thresholds from training-normal scores
train_scores = train_df["anomaly_score"].to_numpy()

WARNING_THRESHOLD = float(np.quantile(train_scores, 0.02))
CRITICAL_THRESHOLD = float(np.quantile(train_scores, 0.005))
CRITICAL_THRESHOLD = min(CRITICAL_THRESHOLD, WARNING_THRESHOLD - 1e-9)

def production_action(score):
    if score < CRITICAL_THRESHOLD:
        return "kill"
    if score < WARNING_THRESHOLD:
        return "isolate"
    return "log"

test_df["production_action"] = test_df["anomaly_score"].apply(production_action)

print(f"WARNING_THRESHOLD  = {WARNING_THRESHOLD:.6f}")
print(f"CRITICAL_THRESHOLD = {CRITICAL_THRESHOLD:.6f}")
```

```python
# 12. Evaluation: never convert unknown into benign
eval_df = test_df.copy()
known = eval_df[eval_df["ground_truth"].isin(["attack", "benign"])].copy()

if known.empty:
    print("No confirmed attack/benign held-out windows.")
else:
    from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix

    known["y_true"] = (known["ground_truth"] == "attack").astype(int)
    known["y_pred"] = known["production_action"].ne("log").astype(int)

    print("Confirmed windows:", len(known))
    print("Attack:", int(known["y_true"].sum()))
    print("Benign:", int((known["y_true"] == 0).sum()))
    print("Precision:", precision_score(known["y_true"], known["y_pred"], zero_division=0))
    print("Recall:   ", recall_score(known["y_true"], known["y_pred"], zero_division=0))
    print("F1:       ", f1_score(known["y_true"], known["y_pred"], zero_division=0))
    print("Confusion matrix [benign, attack] x [log, action]:")
    print(confusion_matrix(known["y_true"], known["y_pred"], labels=[0, 1]))

attack_eval = eval_df[eval_df["ground_truth"] == "attack"].copy()
if not attack_eval.empty:
    print("\nAttack-window detection rate:",
          round(float(attack_eval["production_action"].ne("log").mean()), 4))
    print(attack_eval["production_action"].value_counts())
```

```python
# 13. Inspect the previously broken 09:48 test-pod-apt label
problem = eval_df[
    eval_df["window_start"].eq(pd.Timestamp("2026-07-24 09:48:00+00:00"))
    & eval_df["pod_name"].eq("test-pod-apt")
]

display(problem[[
    "window_start", "pod_name", "event_count",
    "anomaly_score", "production_action", "ground_truth"
]])

if not problem.empty:
    assert problem.iloc[0]["ground_truth"] == "attack"
```

```python
# 14. Save the trained model and complete feature contract
import joblib

bundle = {
    "model": model,
    "feature_columns": FEATURE_COLUMNS,
    "rule_features": RULE_FEATURES,
    "security_rules": SECURITY_RULES,
    "window_seconds": WINDOW_SECONDS,
    "priority_weight": PRIORITY_WEIGHT,
    "warning_threshold": WARNING_THRESHOLD,
    "critical_threshold": CRITICAL_THRESHOLD,
    "excluded_namespaces": sorted(EXCLUDED_NAMESPACES),
    "version": "falco-isolation-forest-fixed-v2",
}

joblib.dump(bundle, DATA_DIR / "falco_isolation_forest_fixed.joblib")
feature_df.to_csv(DATA_DIR / "feature_windows_fixed.csv", index=False)
eval_df.to_csv(DATA_DIR / "test_predictions_fixed.csv", index=False)

print("Saved falco_isolation_forest_fixed.joblib")
print("Saved feature_windows_fixed.csv")
print("Saved test_predictions_fixed.csv")
```

## Runtime scoring contract

The production remediation engine must use the same feature columns and rule vocabulary saved in the bundle. Do not keep the old five-rule `KNOWN_RULES` list in the runtime extractor.


```python
# 15. Runtime-compatible scoring helper
def score_live_window(events_for_pod, pod_name, window_start, bundle):
    live = events_for_pod.copy()
    live["time"] = pd.to_datetime(live["time"], utc=True, errors="coerce")
    live["pod_name"] = live["pod_name"].astype(str)
    live["priority"] = live["priority"].fillna("Unknown")
    live["rule"] = live["rule"].fillna("Unknown")
    live["priority_weight"] = live["priority"].map(
        bundle["priority_weight"]
    ).fillna(0.0)

    row = build_window(
        live,
        pd.Timestamp(window_start),
        pod_name,
        rule_features=bundle["rule_features"],
    )

    X = pd.DataFrame([row])[bundle["feature_columns"]].astype(float).fillna(0.0)
    score = float(bundle["model"].decision_function(X)[0])

    if score < bundle["critical_threshold"]:
        action = "kill"
    elif score < bundle["warning_threshold"]:
        action = "isolate"
    else:
        action = "log"

    return score, action, row
```

```python
# 16. Final sanity checks
assert set(TEST_PODS).isdisjoint(set(train_df["pod_name"]))
assert WARNING_THRESHOLD > CRITICAL_THRESHOLD
assert len(FEATURE_COLUMNS) > 20
assert "rule_Launch" not in FEATURE_COLUMNS  # no accidental unstructured names
print("SANITY CHECKS PASSED")
print("Model:", type(model).__name__)
print("Training rows:", len(train_df))
print("Held-out rows:", len(test_df))
print("Features:", len(FEATURE_COLUMNS))
print("Warning threshold:", WARNING_THRESHOLD)
print("Critical threshold:", CRITICAL_THRESHOLD)
```
