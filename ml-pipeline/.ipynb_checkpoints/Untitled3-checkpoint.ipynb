```python
import json
import os
import re
import warnings
from collections import Counter

import joblib
import numpy as np
import pandas as pd

from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
)

warnings.filterwarnings("ignore")

# ============================================================
# CONFIGURATION
# ============================================================

DATA_FILE = "falco_alerts.jsonl"
GROUND_TRUTH_FILE = "ground_truth_log.csv"

WINDOW_SECONDS = 120

RANDOM_STATE = 42

# Initial model configuration.
# We will tune these later using validation data.
CONTAMINATION = 0.10
N_ESTIMATORS = 200

print("V2 feature engineering environment ready.")
print("Window:", WINDOW_SECONDS, "seconds")
print("Contamination:", CONTAMINATION)
print("Estimators:", N_ESTIMATORS)
```

```python
events = []

with open(DATA_FILE, "r") as f:
    for line in f:
        if not line.strip():
            continue

        event = json.loads(line)

        fields = event.get("output_fields", {})

        events.append({
            "uuid": event.get("uuid"),
            "time": event.get("time"),
            "rule": event.get("rule"),
            "priority": event.get("priority"),
            "output": event.get("output"),
            "source": event.get("source"),
            "tags": event.get("tags"),

            # Kubernetes
            "pod": fields.get("k8s.pod.name"),
            "namespace": fields.get("k8s.ns.name"),

            # Container
            "container_id": fields.get("container.id"),
            "container_name": fields.get("container.name"),
            "image": fields.get("container.image.repository"),
            "image_tag": fields.get("container.image.tag"),
            "container_mounts": fields.get("container.mounts"),
            "container_start_ts": fields.get("container.start_ts"),

            # Process
            "proc_name": fields.get("proc.name"),
            "proc_exe": fields.get("proc.exe"),
            "proc_exepath": fields.get("proc.exepath"),
            "proc_cmdline": fields.get("proc.cmdline"),
            "proc_pname": fields.get("proc.pname"),
            "proc_pcmdline": fields.get("proc.pcmdline"),
            "proc_cwd": fields.get("proc.cwd"),
            "proc_env": fields.get("proc.env"),

            # Event
            "evt_type": fields.get("evt.type"),
            "evt_res": fields.get("evt.res"),
            "evt_arg_name": fields.get("evt.arg.name"),
            "evt_arg_path": fields.get("evt.arg.path"),
            "evt_arg_oldpath": fields.get("evt.arg.oldpath"),
            "evt_arg_filename": fields.get("evt.arg.filename"),
            "evt_arg_flags": fields.get("evt.arg.flags"),
            "evt_arg_mode": fields.get("evt.arg.mode"),

            # File descriptor / network
            "fd_name": fields.get("fd.name"),
            "fd_type": fields.get("fd.type"),
            "l4proto": fields.get("fd.l4proto"),
            "src_ip": fields.get("fd.sip"),
            "src_port": fields.get("fd.lport"),
            "dst_port": fields.get("fd.rport"),

            # Identity
            "uid": fields.get("user.uid"),
            "username": fields.get("user.name"),
            "loginuid": fields.get("user.loginuid"),

            # Falco health counters
            "n_evts": fields.get("n_evts"),
            "n_drops": fields.get("n_drops"),
        })

df = pd.DataFrame(events)

df["time"] = pd.to_datetime(df["time"], utc=True, errors="coerce")

df = df.sort_values("time").reset_index(drop=True)

print("Total events:", len(df))
print("Columns:", len(df.columns))
print("\nTime range:")
print(df["time"].min())
print(df["time"].max())

print("\nPriority distribution:")
print(df["priority"].value_counts(dropna=False))

print("\nRule count:")
print(df["rule"].value_counts().head(30))
```

```python
before = len(df)

if "uuid" in df.columns:
    df = df.drop_duplicates(subset=["uuid"], keep="first")

after = len(df)

print("Events before UUID deduplication:", before)
print("Events after UUID deduplication:", after)
print("Removed:", before - after)
```

```python
def classify_context(row):
    pod = row["pod"]
    container_id = row["container_id"]
    container_name = row["container_name"]
    proc = row["proc_name"]
    uid = row["uid"]

    # Strong Kubernetes attribution
    if pd.notna(pod):
        return "container"

    # Container runtime context even without pod metadata
    if pd.notna(container_id) and str(container_name).lower() != "host":
        return "container"

    # Host / low-context activity
    if str(container_name).lower() == "host":
        if pd.isna(proc) or str(uid) == "4294967295":
            return "host_low_context"
        return "host"

    # We still have process/container evidence
    if pd.notna(proc) or pd.notna(container_id):
        return "runtime_unknown"

    return "low_context"


df["telemetry_context"] = df.apply(classify_context, axis=1)

print(df["telemetry_context"].value_counts())

print("\nContext × priority:")
print(
    pd.crosstab(
        df["telemetry_context"],
        df["priority"]
    )
)
```

```python
TEXT_COLUMNS = [
    "rule",
    "proc_name",
    "proc_exe",
    "proc_exepath",
    "proc_cmdline",
    "proc_pname",
    "proc_pcmdline",
    "evt_arg_name",
    "evt_arg_path",
    "evt_arg_oldpath",
    "evt_arg_filename",
    "fd_name",
    "container_name",
    "image",
]

for col in TEXT_COLUMNS:
    if col in df.columns:
        df[col] = df[col].fillna("").astype(str).str.lower()

print("Text normalization complete.")
```

```python
def contains_any(series, patterns):
    pattern = "|".join(re.escape(x) for x in patterns)
    return series.str.contains(pattern, regex=True, na=False)


# ------------------------------------------------------------
# Process / execution
# ------------------------------------------------------------

df["is_exec"] = (df["evt_type"] == "execve").astype(int)

df["is_shell"] = contains_any(
    df["proc_name"],
    ["sh", "bash", "zsh", "dash", "ksh", "fish"]
).astype(int)

df["is_download_tool"] = contains_any(
    df["proc_name"],
    ["wget", "curl", "ftp", "tftp", "rsync"]
).astype(int)

df["is_package_manager"] = contains_any(
    df["proc_name"],
    ["apt", "apt-get", "yum", "dnf", "apk", "rpm", "pip", "npm"]
).astype(int)

df["is_interpreter"] = contains_any(
    df["proc_name"],
    ["python", "python3", "perl", "ruby", "php", "node"]
).astype(int)

# ------------------------------------------------------------
# Filesystem
# ------------------------------------------------------------

df["is_write_event"] = contains_any(
    df["evt_type"],
    ["write", "pwrite", "openat", "openat2", "creat"]
).astype(int)

df["is_delete_event"] = contains_any(
    df["evt_type"],
    ["unlink", "unlinkat", "rename", "renameat"]
).astype(int)

df["is_chmod_event"] = contains_any(
    df["evt_type"],
    ["chmod", "fchmod", "fchmodat"]
).astype(int)

# ------------------------------------------------------------
# Sensitive files
# ------------------------------------------------------------

df["touches_sensitive_file"] = contains_any(
    df[
        [
            "evt_arg_path",
            "evt_arg_name",
            "evt_arg_filename",
            "fd_name",
        ]
    ].fillna("").agg(" ".join, axis=1),
    [
        "/etc/shadow",
        "/etc/passwd",
        ".ssh",
        "authorized_keys",
        "id_rsa",
        "known_hosts",
        ".bash_history",
        ".zsh_history",
        "/etc/sudoers",
    ],
).astype(int)

# ------------------------------------------------------------
# Network
# ------------------------------------------------------------

df["is_network_event"] = (
    df["evt_type"].isin(["connect", "sendto", "recvfrom", "accept", "accept4"])
).astype(int)

df["is_udp"] = (
    df["l4proto"].astype(str).str.lower() == "udp"
).astype(int)

# ------------------------------------------------------------
# Kubernetes API
# ------------------------------------------------------------

df["is_k8s_api"] = (
    df["rule"] == "contact k8s api server from container"
).astype(int)

# ------------------------------------------------------------
# Privilege
# ------------------------------------------------------------

df["is_privilege_rule"] = contains_any(
    df["rule"],
    [
        "setuid",
        "setgid",
        "privilege",
        "namespace privileges",
        "suid",
    ],
).astype(int)

# ------------------------------------------------------------
# Container / mount
# ------------------------------------------------------------

df["is_sensitive_mount"] = (
    df["rule"] == "launch sensitive mount container"
).astype(int)

# ------------------------------------------------------------
# Rule-level indicators
# ------------------------------------------------------------

df["is_binary_drop"] = (
    df["rule"] == "drop and execute new binary in container"
).astype(int)

df["is_write_root"] = (
    df["rule"] == "write below root"
).astype(int)

df["is_ssh_access"] = (
    df["rule"] == "read ssh information"
).astype(int)

df["is_history_delete"] = (
    df["rule"] == "delete or rename shell history"
).astype(int)

df["is_env_privilege"] = (
    df["rule"] == "potential local privilege escalation via environment variables misuse"
).astype(int)

df["is_namespace_change"] = contains_any(
    df["rule"],
    ["change thread namespace", "change namespace privileges via unshare"]
).astype(int)

print("Behavioral indicators created.")

indicator_cols = [
    "is_exec",
    "is_shell",
    "is_download_tool",
    "is_package_manager",
    "is_interpreter",
    "is_write_event",
    "is_delete_event",
    "is_chmod_event",
    "touches_sensitive_file",
    "is_network_event",
    "is_udp",
    "is_k8s_api",
    "is_privilege_rule",
    "is_sensitive_mount",
    "is_binary_drop",
    "is_write_root",
    "is_ssh_access",
    "is_history_delete",
    "is_env_privilege",
    "is_namespace_change",
]

print("\nIndicator totals:")
print(df[indicator_cols].sum().sort_values(ascending=False))
```

```python
behavior_text = (
    df[
        [
            "proc_cmdline",
            "proc_pcmdline",
            "evt_arg_path",
            "evt_arg_name",
            "evt_arg_filename",
            "fd_name",
        ]
    ]
    .fillna("")
    .agg(" ".join, axis=1)
)

df["cmd_download"] = contains_any(
    behavior_text,
    ["wget ", "curl ", "ftp ", "tftp "]
).astype(int)

df["cmd_shell_execution"] = contains_any(
    behavior_text,
    [
        "/bin/sh",
        "/bin/bash",
        " sh -c ",
        "bash -c",
        "exec /bin/sh",
        "exec /bin/bash",
    ]
).astype(int)

df["cmd_chmod"] = contains_any(
    behavior_text,
    ["chmod ", "chmod +", "chown "]
).astype(int)

df["cmd_network_tool"] = contains_any(
    behavior_text,
    ["nc ", "netcat", "socat", "curl ", "wget "]
).astype(int)

df["cmd_sensitive_file"] = contains_any(
    behavior_text,
    [
        "/etc/shadow",
        "/etc/passwd",
        "/root/.ssh",
        "authorized_keys",
        "id_rsa",
        ".bash_history",
    ],
).astype(int)

df["cmd_write_root"] = contains_any(
    behavior_text,
    [
        "/etc/",
        "/root/",
        "/usr/bin/",
        "/usr/local/bin/",
        "/bin/",
        "/sbin/",
    ],
).astype(int)

print("Command/path indicators created.")
```

```python
# Prefer container ID because it remains meaningful even when pod metadata
# is missing. Fall back to pod/container name where necessary.

df["behavior_key"] = (
    df["container_id"]
    .replace("", np.nan)
    .fillna(df["pod"])
    .fillna(df["container_name"])
    .fillna("unknown")
)

df = df.sort_values("time").reset_index(drop=True)

print("Unique behavioral entities:", df["behavior_key"].nunique())
print("Total events:", len(df))
```

```python
PRIORITY_WEIGHT = {
    "Critical": 4,
    "Error": 3,
    "Warning": 2,
    "Notice": 1,
    "Informational": 0,
    "Debug": 0,
}


def build_window_features(group):
    """
    Convert all events belonging to one behavioral entity into
    non-overlapping 120-second behavioral windows.
    """

    group = group.sort_values("time").copy()

    start = group["time"].min()
    elapsed = (group["time"] - start).dt.total_seconds()

    group["window_id"] = (
        np.floor(elapsed / WINDOW_SECONDS).astype(int)
    )

    rows = []

    for window_id, w in group.groupby("window_id"):

        if len(w) == 0:
            continue

        rules = set(w["rule"])
        processes = set(x for x in w["proc_name"] if x)
        event_types = set(x for x in w["evt_type"] if x)

        weights = [
            PRIORITY_WEIGHT.get(x, 0)
            for x in w["priority"]
        ]

        row = {
            # Identity
            "behavior_key": w["behavior_key"].iloc[0],
            "window_id": window_id,
            "window_start": w["time"].min(),
            "window_end": w["time"].max(),

            # Volume
            "event_count": len(w),
            "events_per_second": len(w) / WINDOW_SECONDS,

            # Diversity
            "distinct_rules": len(rules),
            "distinct_processes": len(processes),
            "distinct_event_types": len(event_types),

            # Priority
            "critical_count": int((w["priority"] == "Critical").sum()),
            "error_count": int((w["priority"] == "Error").sum()),
            "warning_count": int((w["priority"] == "Warning").sum()),
            "notice_count": int((w["priority"] == "Notice").sum()),
            "max_priority_weight": max(weights) if weights else 0,
            "avg_priority_weight": np.mean(weights) if weights else 0,

            # Context quality
            "container_event_count": int(
                (w["telemetry_context"] == "container").sum()
            ),
            "host_low_context_count": int(
                (w["telemetry_context"] == "host_low_context").sum()
            ),

            # Process
            "exec_count": w["is_exec"].sum(),
            "shell_count": w["is_shell"].sum(),
            "download_tool_count": w["is_download_tool"].sum(),
            "package_manager_count": w["is_package_manager"].sum(),
            "interpreter_count": w["is_interpreter"].sum(),

            # Filesystem
            "write_event_count": w["is_write_event"].sum(),
            "delete_event_count": w["is_delete_event"].sum(),
            "chmod_event_count": w["is_chmod_event"].sum(),
            "sensitive_file_count": w["touches_sensitive_file"].sum(),

            # Network
            "network_event_count": w["is_network_event"].sum(),
            "udp_event_count": w["is_udp"].sum(),
            "unique_dst_ports": w["dst_port"].dropna().nunique(),

            # Identity / privilege
            "privilege_event_count": w["is_privilege_rule"].sum(),

            # Kubernetes
            "k8s_api_count": w["is_k8s_api"].sum(),

            # Specific security behaviors
            "binary_drop_count": w["is_binary_drop"].sum(),
            "write_root_count": w["is_write_root"].sum(),
            "ssh_access_count": w["is_ssh_access"].sum(),
            "history_delete_count": w["is_history_delete"].sum(),
            "env_privilege_count": w["is_env_privilege"].sum(),
            "namespace_change_count": w["is_namespace_change"].sum(),
            "sensitive_mount_count": w["is_sensitive_mount"].sum(),

            # Command/path context
            "cmd_download_count": w["cmd_download"].sum(),
            "cmd_shell_count": w["cmd_shell_execution"].sum(),
            "cmd_network_tool_count": w["cmd_network_tool"].sum(),
            "cmd_sensitive_file_count": w["cmd_sensitive_file"].sum(),
            "cmd_write_root_count": w["cmd_write_root"].sum(),
            "cmd_chmod_count": w["cmd_chmod"].sum(),

            # Temporal correlation
            "download_then_exec": int(
                w["is_download_tool"].sum() > 0
                and w["is_exec"].sum() > 0
            ),

            "exec_then_privilege": int(
                w["is_exec"].sum() > 0
                and w["is_privilege_rule"].sum() > 0
            ),

            "exec_then_sensitive_file": int(
                w["is_exec"].sum() > 0
                and w["touches_sensitive_file"].sum() > 0
            ),

            "network_then_binary_drop": int(
                w["is_network_event"].sum() > 0
                and w["is_binary_drop"].sum() > 0
            ),

            "shell_then_sensitive_file": int(
                w["is_shell"].sum() > 0
                and w["touches_sensitive_file"].sum() > 0
            ),

            "k8s_api_then_exec": int(
                w["is_k8s_api"].sum() > 0
                and w["is_exec"].sum() > 0
            ),
        }

        rows.append(row)

    return pd.DataFrame(rows)


window_frames = []

for key, group in df.groupby("behavior_key", dropna=False):
    result = build_window_features(group)
    window_frames.append(result)

features_df = pd.concat(
    window_frames,
    ignore_index=True
)

print("Behavioral windows:", len(features_df))
print("Features:", len(features_df.columns))

features_df.head()
```

```python
NON_FEATURE_COLUMNS = [
    "behavior_key",
    "window_id",
    "window_start",
    "window_end",
]

FEATURE_COLUMNS = [
    c for c in features_df.columns
    if c not in NON_FEATURE_COLUMNS
]

print("Number of ML features:", len(FEATURE_COLUMNS))

print("\nML features:")
for i, col in enumerate(FEATURE_COLUMNS, 1):
    print(f"{i:02d}. {col}")
```

```python
feature_summary = pd.DataFrame({
    "feature": FEATURE_COLUMNS,
    "nonzero_windows": [
        int((features_df[c] > 0).sum())
        for c in FEATURE_COLUMNS
    ],
    "mean": [
        features_df[c].mean()
        for c in FEATURE_COLUMNS
    ],
    "max": [
        features_df[c].max()
        for c in FEATURE_COLUMNS
    ],
})

print(feature_summary.to_string(index=False))
```
